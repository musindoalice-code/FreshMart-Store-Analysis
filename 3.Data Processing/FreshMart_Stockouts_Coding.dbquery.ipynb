```sql
SELECT * 
FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1
 LIMIT 5;

 --Checking the date range
 SELECT 
 MIN(date) AS earliest, 
 MAX(date) AS latest 
 FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1;

 --checking how many stores are affected
 SELECT COUNT(DISTINCT store_id) AS n_stores_affected
  FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1;

--checking which categories where affected
SELECT category, COUNT(*) AS n_events
FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1
GROUP BY category
ORDER BY n_events DESC;

--checking how long stockouts last
SELECT 
MIN(duration_hours) AS min_hrs, 
MAX(duration_hours) AS max_hrs, 
ROUND(AVG(duration_hours),1) AS avg_hrs
FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1;

--checking which stores are affected too much
SELECT store_id, COUNT(*) AS n_events
FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1
GROUP BY store_id
ORDER BY n_events DESC;

--checking when the stockouts happen
SELECT date_format(CAST(date AS DATE), 'EEEE') AS weekday, COUNT(*) AS n_events
FROM freshmart_analysis.default.fresh_mart_stockoutsdataset_1
GROUP BY weekday
ORDER BY n_events DESC;
```
