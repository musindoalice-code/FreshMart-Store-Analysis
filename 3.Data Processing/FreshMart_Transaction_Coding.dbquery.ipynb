```sql

--Checking how big is the dataset
SELECT COUNT(*) AS total_rows 
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1;

--Checking what the data looks like
SELECT * 
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
LIMIT 5;

--Checking the date range
SELECT 
MIN(transaction_date) AS earliest,
 MAX(transaction_date) AS latest
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1;

--checking how many stores there are
SELECT COUNT(DISTINCT store_id) AS num_stores 
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1;

--Checking Missing Values
SELECT
    SUM(CASE WHEN basket_value_zar IS NULL THEN 1 ELSE 0 END) AS missing_basket_value,
    SUM(CASE WHEN num_items IS NULL THEN 1 ELSE 0 END) AS missing_num_items
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1;

--Checking Loyalty Members and non loyalty members
SELECT is_loyalty_member, COUNT(*) AS n
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY is_loyalty_member;

--Checking the average of the basket 
SELECT ROUND(AVG(basket_value_zar),2) AS avg_basket_value,
       ROUND(AVG(num_items),2) AS avg_items
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1;

--Checking min,max values
SELECT MIN(basket_value_zar) AS min_basket, 
       MAX(basket_value_zar) AS max_basket,
       MIN(num_items) AS min_items, 
       MAX(num_items) AS max_items
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1;

--checking payment methods
SELECT payment_method, COUNT(*) AS n
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY payment_method
ORDER BY n DESC;

--checking Day of the week
SELECT day_of_week, ROUND(AVG(basket_value_zar),2) AS avg_basket
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY day_of_week
ORDER BY avg_basket DESC;

--checking the basket fall
SELECT date_format(CAST(transaction_date AS DATE), 
       'yyyy-MM') AS month,
       COUNT(*) AS num_transactions,
       ROUND(AVG(basket_value_zar),2) AS avg_basket_value,
       ROUND(SUM(basket_value_zar),0) AS total_revenue
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY month
ORDER BY month;

--checking quaterly revenues
SELECT 
    CASE 
       WHEN transaction_date < DATE '2025-07-01' THEN 'H1 2025' ELSE 'H2 2025' END AS half,
       COUNT(*) AS num_transactions,
       ROUND(AVG(basket_value_zar),2) AS avg_basket_value,
       ROUND(AVG(num_items),2) AS avg_items
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY half;

--checking Loyalty Status Hypothesis
SELECT is_loyalty_member, 
ROUND(AVG(basket_value_zar),2) AS avg_basket
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY is_loyalty_member;

-- Members check only
SELECT 
    CASE 
    WHEN transaction_date < DATE '2025-07-01' 
    THEN 'H1' 
    ELSE 'H2' 
    END AS half,
    COUNT(*) AS n, 
    ROUND(AVG(basket_value_zar),2) AS avg_basket
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
WHERE is_loyalty_member = True
GROUP BY half;

--Checking the mix of basket change
SELECT 
     CASE 
     WHEN basket_value_zar < 150 THEN 'Small (<R150)'
     WHEN basket_value_zar < 400 THEN 'Medium (R150-399)'
     ELSE 'Large (R400+)' 
     END AS band,
     COUNT(*) AS n
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
WHERE transaction_date < DATE '2025-07-01'   -- run again with >= for H2
GROUP BY band;


SELECT 
     CASE 
     WHEN basket_value_zar < 150 THEN 'Small (<R150)'
     WHEN basket_value_zar < 400 THEN 'Medium (R150-399)'
     ELSE 'Large (R400+)' 
     END AS band,
     COUNT(*) AS n
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
WHERE transaction_date >= DATE '2025-07-01'  
GROUP BY band;

--Checking weekday and weekend analysis
SELECT 
    CASE 
    WHEN day_of_week IN ('Saturday','Sunday') 
    THEN 'Weekend' 
    ELSE 'Weekday' 
    END AS day_type,
    CASE WHEN transaction_date < DATE '2025-07-01' 
    THEN 'H1' 
    ELSE 'H2' 
    END AS half,
    COUNT(*) AS n, ROUND(AVG(basket_value_zar),2) AS avg_basket
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY day_type, half;

--Checking of its really pricing falling
SELECT date_format(CAST(transaction_date AS DATE), 'yyyy-MM') AS month,
       ROUND(AVG(num_items),2) AS avg_items,
       ROUND(AVG(basket_value_zar)/AVG(num_items),2) AS avg_price_per_item
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY month
ORDER BY month;

--Checking Quater analysis
SELECT date_format(CAST(transaction_date AS DATE), 'yyyy-MM') AS month,
       ROUND(AVG(basket_value_zar),2) AS avg_basket
FROM freshmart_analysis.default.fresh_mart_transactiondataset_1
GROUP BY month
HAVING AVG(basket_value_zar) < 270
ORDER BY month;
```
