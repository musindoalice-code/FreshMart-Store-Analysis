```sql
SELECT * 
FROM freshmart_analysis.default.fresh_mart_storesdataset_1
LIMIT 5;

--Checking province stores
SELECT province, COUNT(*) AS n_stores
FROM freshmart_analysis.default.fresh_mart_storesdataset_1
GROUP BY province
ORDER BY n_stores DESC;

--checking store formats
SELECT store_format, COUNT(*) AS n_stores
FROM freshmart_analysis.default.fresh_mart_storesdataset_1
GROUP BY store_format
ORDER BY n_stores DESC;

--Checking floor area
SELECT store_format, 
ROUND(AVG(floor_area_sqm),0) AS avg_area,
       MIN(floor_area_sqm) AS min_area, MAX(floor_area_sqm) AS max_area
FROM freshmart_analysis.default.fresh_mart_storesdataset_1
GROUP BY store_format
ORDER BY avg_area DESC;

--checking stores with competitors
SELECT has_nearby_competitor, COUNT(*) AS n_stores
FROM freshmart_analysis.default.fresh_mart_storesdataset_1
GROUP BY has_nearby_competitor;

--checking open dates for competitor stores
SELECT store_id, store_name, province, store_format, competitor_open_date
FROM freshmart_analysis.default.fresh_mart_storesdataset_1
WHERE has_nearby_competitor = 'Yes'
ORDER BY competitor_open_date;
```
